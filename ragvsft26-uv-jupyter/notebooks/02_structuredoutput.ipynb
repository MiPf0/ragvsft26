```python
import ollama
from pydantic import BaseModel

# Define your structure
class User(BaseModel):
    name: str
    age: int
    hobbies: list[str]

response = ollama.chat(
    model='llama3.1',
    messages=[{'role': 'user', 'content': 'Extract: Harry (38 years old) loves to knit. He also likes cooking.'}],
    # This is the magic part
    format=User.model_json_schema(), 
)

print(response.message.content)
```

```python
from typing import Literal
from pydantic import BaseModel
import ollama

class Classification(BaseModel):
    sentiment: Literal["Positive", "Negative", "Neutral"]
    topic: Literal["Tech", "Health", "Finance", "Other"]
    confidence_score: float

response = ollama.chat(
    model='llama3.1',
    messages=[{'role': 'user', 'content': 'I love the new CPU architecture!'}],
    format=Classification.model_json_schema(),
)

print(response.message.content)
```
